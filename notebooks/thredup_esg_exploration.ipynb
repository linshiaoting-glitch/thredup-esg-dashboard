# ThredUp — ESG data exploration (SAM 503 · Lab 4)
Load, clean and explore the data before building the Streamlit dashboard.

```python
import pandas as pd, matplotlib.pyplot as plt
BASE = 'https://raw.githubusercontent.com/USER/thredup-esg-dashboard/main/data/'
try:
    ts = pd.read_csv(BASE + 'thredup_timeseries.csv')
except Exception:
    BASE = 'data/'
    ts = pd.read_csv(BASE + 'thredup_timeseries.csv')
ts[['year','revenue_m_usd','net_loss_m_usd','employees','basis']]
```

## Cleaning: flag reporting basis and missing ESG data

```python
print(ts.isna().sum()[['revenue_m_usd','cum_items_processed_m','cum_co2_avoided_m_lbs','workforce_female_pct']])
# 2025 revenue must match FY2025 results
assert ts.loc[ts.year==2025,'revenue_m_usd'].item() == 310.8
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].bar(ts.year, ts.revenue_m_usd); ax[0].set_title('Revenue ($M)')
ax[1].plot(ts.year, ts.net_margin_pct, marker='o', color='#eb6834'); ax[1].axhline(0, c='grey'); ax[1].set_title('Net loss margin (%)')
plt.tight_layout(); plt.show()
```

## Avoided emissions check — a cumulative number should never fall

```python
av = ts.dropna(subset=['cum_co2_avoided_m_lbs'])[['year','cum_co2_avoided_m_lbs']]
av['change'] = av.cum_co2_avoided_m_lbs.diff()
print(av)
print('Methodology change detected' if (av.change < 0).any() else 'Consistent')
```
